```python
from pathlib import Path
import pandas as pd

# 1. Set the correct project directory
PROJECT_DIR = Path(
    r"C:\Users\Tejas\Desktop\CRISPR_bioinfo_project"
)

# 2. Set the data directory
DATA_DIR = PROJECT_DIR / "data"

# 3. Display the paths
print("PROJECT DIRECTORY:", PROJECT_DIR)
print("DATA DIRECTORY:", DATA_DIR)

# 4. Check whether the data folder exists
if not DATA_DIR.is_dir():
    raise FileNotFoundError(
        f"Data folder not found: {DATA_DIR}"
    )

# 5. Display the files actually present
print("\nFILES IN DATA FOLDER")

for file in sorted(DATA_DIR.iterdir()):
    if file.is_file():
        print(file.name)

# 6. Set the guide file path
GUIDE_FILE = DATA_DIR / "crispor_guides.tsv.xls"

# 7. Check whether the guide file exists
if not GUIDE_FILE.is_file():
    raise FileNotFoundError(
        f"\nGuide file not found: {GUIDE_FILE}\n"
        "Check the filenames printed above and update GUIDE_FILE."
    )

print("\nGuide file found:", GUIDE_FILE.name)

# 8. Read the file according to its actual format
with GUIDE_FILE.open("rb") as file:
    signature = file.read(8)

if signature.startswith(b"\xD0\xCF\x11\xE0"):
    guides_df = pd.read_excel(GUIDE_FILE, engine="xlrd")

elif signature.startswith(b"PK\x03\x04"):
    guides_df = pd.read_excel(GUIDE_FILE, engine="openpyxl")

else:
    guides_df = pd.read_csv(
        GUIDE_FILE,
        sep="\t",
        dtype=str,
        encoding="utf-8"
    )

# 9. Display the dataset
print("\nFIRST 10 ROWS")
print(guides_df.head(10).to_string(index=False))

print("\nCOLUMN NAMES")
print(guides_df.columns.tolist())

print("\nDATASET DIMENSIONS")
print("Rows:", guides_df.shape[0])
print("Columns:", guides_df.shape[1])
```

```python
print("\nMISSING VALUES PER COLUMN")
print(guides_df.isna().sum())

print("\nDUPLICATED ROWS")
print(guides_df.duplicated().sum())
```
