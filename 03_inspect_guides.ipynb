```python
from pathlib import Path
import pandas as pd

# 1. Define the project directory
PROJECT_DIR = Path(r"C:\Users\Tejas\Desktop\CRISPR_bioinfo_project")

# 2. Define the data directory
DATA_DIR = PROJECT_DIR / "data"

# 3. Define the guide file
GUIDE_FILE = DATA_DIR / "crispor_guides.tsv.xls"

# 4. Check whether the file exists
if not GUIDE_FILE.exists():
    raise FileNotFoundError(
        f"Guide file not found: {GUIDE_FILE}\n"
        "Check the filename in your data folder."
    )

print("Guide file found:", GUIDE_FILE.name)

# 5. Read the Excel file
df = pd.read_excel(GUIDE_FILE)

# 6. Display the first 10 rows
print("\nFIRST 10 ROWS")
print("-------------")
print(df.head(10).to_string(index=False))

# 7. Display column names
print("\nCOLUMN NAMES")
print("------------")
print(df.columns.tolist())

# 8. Display dataset dimensions
print("\nDATASET SIZE")
print("------------")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```
