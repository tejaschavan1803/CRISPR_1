```python
from pathlib import Path

PROJECT_DIR = Path(r"C:\Users\Tejas\Desktop\CRISPR_bioinfo_project")
DATA_DIR = PROJECT_DIR / "data"

OFFTARGET_FILE = DATA_DIR / "crispor_offtargets.tsv.xls"

if not OFFTARGET_FILE.exists():
    raise FileNotFoundError(
        f"Off-target file not found: {OFFTARGET_FILE}"
    )

print("Off-target file found:", OFFTARGET_FILE.name)

print("\nFIRST 10 LINES")
print("--------------")

with open(
    OFFTARGET_FILE,
    "r",
    encoding="utf-8",
    errors="replace"
) as file:

    for _ in range(10):
        line = file.readline()

        if not line:
            break

        print(line.rstrip())
```
