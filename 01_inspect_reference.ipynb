```python
from pathlib import Path
from Bio import SeqIO
from collections import Counter

# Locate the project directory
PROJECT_DIR = Path(
    "C:/Users/Tejas/Desktop/CRISPR_bioinfo_project/Data/reference_sequnce.fasta"
).resolve().parent.parent

print("Project directory:", PROJECT_DIR)

## locate the reference file
FASTA_FILE = PROJECT_DIR / "Data" / "reference_sequence.fasta"

# check the file
print("Refernce file:",FASTA_FILE.exists())

#read the FASTA file
record = SeqIO.read(FASTA_FILE, "fasta")

# Display basic information
print("\nREFERENCE INFORMATION")
print("---------------------")
print("Sequence ID:", record.id)
print("Sequence length:", len(record))
print("First 50 bases:", record[:50])


## count nucleotides
base_counts = Counter(record)
print("\nNucleotide counts")

for base in ["A","T","G","C","N"]:
    print(f"{base}:{base_counts.get(base, 0)}")


## check sequence characters
valid_bases = set("ATGCN")
invalid_bases = sorted(set(record)-valid_bases)
print("\nInvalid characters:", invalid_bases)

if not invalid_bases:
    print("sequence character check passed")
else:
    print("Review the unexpected characters")
```
